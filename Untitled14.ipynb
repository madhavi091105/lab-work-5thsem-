```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("computervision/straightline.jpeg")

if img is None:
    img = cv2.imread("straightline.jpeg")

if img is None:
    print("Image not found! Check the image path.")

else:
    # Convert image to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Make sure input is a NumPy array
    gray = np.ascontiguousarray(gray, dtype=np.float32)

    # -------------------------------
    # 1. Harris Corner Detection
    # -------------------------------
    harris = cv2.cornerHarris(gray, 2, 3, 0.04)
    harris = cv2.dilate(harris, None)

    harris_img = img_rgb.copy()
    harris_img[harris > 0.01 * harris.max()] = [255, 0, 0]

    # -------------------------------
    # 2. Shi-Tomasi Corner Detection
    # -------------------------------
    shi_img = img_rgb.copy()

    corners = cv2.goodFeaturesToTrack(
        np.uint8(gray),
        maxCorners=100,
        qualityLevel=0.01,
        minDistance=10
    )

    if corners is not None:
        for corner in corners:
            x, y = corner.ravel().astype(int)
            cv2.circle(shi_img, (x, y), 4, (0, 255, 0), -1)

        print("Shi-Tomasi corners detected:", len(corners))
    else:
        print("No corners detected.")

    # -------------------------------
    # 3. Display results
    # -------------------------------
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(harris_img)
    plt.title("Harris Corner Detection")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(shi_img)
    plt.title("Shi-Tomasi Corner Detection")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python

import cv2
import numpy as np

# Step 1: Load image
img = cv2.imread("computervision/straightline.jpeg")

if img is None:
    img = cv2.imread("straightline.jpeg")

if img is None:
    print("Image not found! Check the file path.")

else:
    # Step 2: Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 3: Harris Corner Detection
    gray_float = np.float32(gray)

    harris = cv2.cornerHarris(
        gray_float,
        blockSize=2,
        ksize=3,
        k=0.04
    )

    harris = cv2.dilate(harris, None)

    harris_img = img.copy()
    harris_img[harris > 0.01 * harris.max()] = [0, 0, 255]

    # Step 4: Shi-Tomasi Corner Detection
    shi_img = img.copy()

    corners = cv2.goodFeaturesToTrack(
        gray,
        maxCorners=100,
        qualityLevel=0.01,
        minDistance=10
    )

    if corners is not None:
        corners = np.int32(corners)

        for corner in corners:
            x, y = corner.ravel()
            cv2.circle(shi_img, (x, y), 5, (0, 255, 0), -1)

        print("Shi-Tomasi corners detected:", len(corners))
    else:
        print("No Shi-Tomasi corners detected.")

    # Step 5: Display results using OpenCV
    cv2.imshow("Original Image", img)
    cv2.imshow("Harris Corner Detection", harris_img)
    cv2.imshow("Shi-Tomasi Corner Detection", shi_img)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
