```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("cd3.jpeg")
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
hsv = cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
```

```python
h, s, v = cv2.split(hsv)
v = cv2.convertScaleAbs(v,alpha = 1.15,beta = 20)
hsv_bright = cv2.merge([h,s,v])
bright_img = cv2.cvtColor(hsv_bright,cv2.COLOR_HSV2BGR)
```

```python
lab = cv2.cvtColor(bright_img, cv2.COLOR_BGR2LAB)

l, a, b = cv2.split(lab)

# CLAHE improves local contrast
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

l = clahe.apply(l)

lab_contrast = cv2.merge([l, a, b])

contrast_img = cv2.cvtColor(
    lab_contrast,
    cv2.COLOR_LAB2BGR
)
```

```python

denoised_img = cv2.fastNlMeansDenoisingColored(
    contrast_img,
    None,
    10,
    10,
    7,
    21
)
```

```python
kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

sharp_img = cv2.filter2D(
    denoised_img,
    -1,
    kernel
)
```

```python
hsv = cv2.cvtColor(sharp_img, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(hsv)

# Increase saturation
s = cv2.convertScaleAbs(
    s,
    alpha=1.20,
    beta=0
)

color_enhanced = cv2.merge([h, s, v])

color_enhanced = cv2.cvtColor(
    color_enhanced,
    cv2.COLOR_HSV2BGR
)
```

```python
gaussian = cv2.GaussianBlur(
    color_enhanced,
    (0,0),
    3
)
final_img = cv2.addWeighted(
    color_enhanced,
    1.5,
    gaussian,
    -0.5,
    0
)

    
    
```

```python
images = [
    img,
    bright_img,
    contrast_img,
    denoised_img,
    sharp_img,
    color_enhanced,
    
]

titles = [
    "Original",
    "Brightness Improved",
    "Contrast Improved",
    "Noise Reduced",
    "Sharpened",
    "Color Enhanced",
    
]
```

```python
plt.figure(figsize=(15,10))
for i in range(len(images)):
    plt.subplot(2,4,i+1)
    plt.imshow(
         cv2.cvtColor(images[i],cv2.COLOR_BGR2RGB)
    )
    plt.title(titles[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
