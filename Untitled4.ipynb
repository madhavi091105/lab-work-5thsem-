```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("cd3.jpeg", cv2.IMREAD_GRAYSCALE)

# -------------------------------
# Add Gaussian Noise
# -------------------------------
mean = 0
sigma = 25

gaussian_noise = np.random.normal(mean, sigma, img.shape)
gaussian_noisy = img.astype(np.float32) + gaussian_noise
gaussian_noisy = np.clip(gaussian_noisy, 0, 255).astype(np.uint8)

# -------------------------------
# Add Salt-and-Pepper Noise
# -------------------------------
sp_noisy = img.copy()

prob = 0.05

# Salt noise
salt = np.random.rand(*img.shape) < prob / 2
sp_noisy[salt] = 255

# Pepper noise
pepper = np.random.rand(*img.shape) < prob / 2
sp_noisy[pepper] = 0

# -------------------------------
# Filtering Gaussian Noise
# -------------------------------

# Mean filter
gaussian_mean = cv2.blur(gaussian_noisy, (5, 5))

# Gaussian filter
gaussian_filter = cv2.GaussianBlur(gaussian_noisy, (5, 5), 0)

# -------------------------------
# Filtering Salt-and-Pepper Noise
# -------------------------------

# Mean filter
sp_mean = cv2.blur(sp_noisy, (5, 5))

# Median filter
sp_median = cv2.medianBlur(sp_noisy, 5)

# -------------------------------
# Display Results
# -------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 4, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(gaussian_noisy, cmap='gray')
plt.title("Gaussian Noise")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(gaussian_mean, cmap='gray')
plt.title("Mean Filter")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(gaussian_filter, cmap='gray')
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(sp_noisy, cmap='gray')
plt.title("Salt & Pepper")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(sp_mean, cmap='gray')
plt.title("Mean Filter")
plt.axis("off")

plt.subplot(2, 4, 8)
plt.imshow(sp_median, cmap='gray')
plt.title("Median Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread("cd3.jpeg", cv2.IMREAD_GRAYSCALE)

# Add salt-and-pepper noise
noisy = img.copy()

noise_probability = 0.08

random_matrix = np.random.rand(*img.shape)

noisy[random_matrix < noise_probability / 2] = 0
noisy[random_matrix > 1 - noise_probability / 2] = 255

# Same kernel size = 5 x 5

# Mean filter
mean_filtered = cv2.blur(noisy, (5, 5))

# Gaussian filter
gaussian_filtered = cv2.GaussianBlur(noisy, (5, 5), 0)

# Median filter
median_filtered = cv2.medianBlur(noisy, 5)

# Display
plt.figure(figsize=(15, 8))

plt.subplot(1, 5, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(noisy, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(mean_filtered, cmap='gray')
plt.title("Mean 5x5")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(gaussian_filtered, cmap='gray')
plt.title("Gaussian 5x5")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(median_filtered, cmap='gray')
plt.title("Median 5x5")
plt.axis("off")

plt.tight_layout()
plt.show()
```
