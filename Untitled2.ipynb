```python
from scipy.stats import poisson
lam = 25
prob = poisson.sf(30,lam)
print("Probability =",prob)
```

```python
# Q1(b)

lam_fraud = 25 * 0.05

prob = 1 - poisson.pmf(0, lam_fraud) - poisson.pmf(1, lam_fraud)

print("Probability of at least 2 fraudulent returns =", prob)
```

```python

n = 8
p = 0.42

prob = binom.sf(2, n, p)

print("Probability of at least 3 calls =", prob)
```

```python

n = 8
p = 0.42

prob = binom.pmf(4, n, p)

print("Probability of exactly 4 calls =", prob)
```

```python

mean = 85

prob = expon.cdf(85, scale=mean)

print("Probability that machine fails before 85 hours =", prob)
```

```python

mean = 85

prob = expon.sf(150, scale=mean)

print("Probability that machine does not fail up to 150 hours =", prob)
```

```python

mean = 38
std = 5

prob = norm.sf(50, loc=mean, scale=std)

print("Proportion spraying more than 50 liters =", prob)
```

```python

mean = 38
std = 5

prob = norm.cdf(10, loc=mean, scale=std)

print("Proportion spraying less than 10 liters =", prob)
```

```python

mean = 38
std = 5

prob = norm.cdf(60, loc=mean, scale=std) - norm.cdf(30, loc=mean, scale=std)

print("Proportion spraying between 30 and 60 liters =", prob)
```

```python
from scipy.stats import norm
import math

n = 60
sample_mean = 4998.1
population_mean = 5000
population_std = 1.30
alpha = 0.05

z = (sample_mean - population_mean) / (population_std / math.sqrt(n))

p_value = norm.cdf(z)

print("Z-statistic =", z)
print("P-value =", p_value)

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
from scipy.stats import norm
import numpy as np

n = 60
sample_mean = 4998.1
population_mean = 5000
population_std = 1.30
alpha = 0.05

z = (sample_mean - population_mean) / (population_std / np.sqrt(n))

p_value = norm.cdf(z)

print("Z-statistic =", z)
print("P-value =", p_value)

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
