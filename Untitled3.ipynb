```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("bollywood.csv")
```

```python
print("Number of records:", len(df))
print("\nMetadata:")
df.info()
```

```python
genre_count = df['Genre'].value_counts().sort_values(ascending=False)

print(genre_count)

print("\nGenre with highest releases:")
print(genre_count.idxmax())
```

```python
cross_tab = pd.crosstab(df['Genre'], df['ReleaseTime'])

print(cross_tab)
```

```python
df['ReleaseDate'] = pd.to_datetime(df['ReleaseDate'])

df['month'] = df['ReleaseDate'].dt.month

month_count = df['month'].value_counts().sort_values(ascending=False)

print(month_count)

print("\nMonth with maximum releases:")
print(month_count.idxmax())
```

```python
high_budget = df[df['Budget'] >= 25]

month_high_budget = high_budget['month'].value_counts().sort_values(ascending=False)

print(month_high_budget)

print("\nMonth with maximum high-budget releases:")
print(month_high_budget.idxmax())
```

```python
df['ROI'] = (df['BoxOfficeCollection'] - df['Budget']) / df['Budget']

top10_roi = df[['MovieName', 'Budget', 'BoxOfficeCollection', 'ROI']] \
              .sort_values(by='ROI', ascending=False) \
              .head(10)

print(top10_roi)
```

```python
avg_roi = df.groupby('ReleaseTime')['ROI'].mean().sort_values(ascending=False)

print(avg_roi)
```

```python
print(df.groupby('ReleaseTime')['ROI'].mean()[['FS', 'LW']])
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Budget'], bins=20)
plt.xlabel("Budget")
plt.ylabel("Number of Movies")
plt.title("Distribution of Movie Budgets")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Budget'], fill=True)
plt.xlabel("Budget")
plt.title("Distribution Plot of Movie Budgets")
plt.show()
```

```python
comedy_roi = df[df['Genre'] == 'Comedy']['ROI']
drama_roi = df[df['Genre'] == 'Drama']['ROI']

print("Average Comedy ROI:", comedy_roi.mean())
print("Average Drama ROI:", drama_roi.mean())
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df[df['Genre'].isin(['Comedy', 'Drama'])],
    x='Genre',
    y='ROI'
)

plt.title("ROI Distribution: Comedy vs Drama")
plt.show()
```

```python
correlation = df['BoxOfficeCollection'].corr(df['YoutubeLikes'])

print("Correlation =", correlation)

if correlation > 0:
    print("Positive correlation")
elif correlation < 0:
    print("Negative correlation")
else:
    print("No correlation")
```

```python
avg_likes = df.groupby('Genre')['YoutubeLikes'].mean().sort_values(ascending=False)

print(avg_likes)

print("\nGenre with highest average YouTube likes:")
print(avg_likes.idxmax())
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x='Genre',
    y='YoutubeLikes'
)

plt.xticks(rotation=45)
plt.title("YouTube Likes by Genre")
plt.show()
```

```python
columns = [
    'Budget',
    'BoxOfficeCollection',
    'YoutubeViews',
    'YoutubeLikes',
    'YoutubeDislikes'
]

corr = df[columns].corr()

print(corr)
```

```python
plt.figure(figsize=(9,7))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
corr_matrix = df[columns].corr()

corr_pairs = corr_matrix.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) != corr_pairs.index.get_level_values(1)
]

print(corr_pairs.abs().sort_values(ascending=False).head(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("SAheart.csv")

print("Number of records:", len(df))

print("\nMetadata:")
df.info()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x="chd", hue="famhist")

plt.xlabel("CHD")
plt.ylabel("Number of Persons")
plt.title("CHD vs Family History")
plt.show()
```

```python
sns.scatterplot(data=df, x="age", y="sbp")

plt.xlabel("Age")
plt.ylabel("SBP")
plt.title("Relationship between Age and SBP")
plt.show()
```

```python
sns.regplot(data=df, x="age", y="sbp")

plt.xlabel("Age")
plt.ylabel("SBP")
plt.title("Correlation between Age and SBP")
plt.show()
```

```python
print("Correlation between age and sbp:",
      df["age"].corr(df["sbp"]))
```

```python
sns.boxplot(data=df, x="chd", y="tobacco")

plt.xlabel("CHD")
plt.ylabel("Tobacco Consumption")
plt.title("Tobacco Consumption for CHD and Non-CHD Persons")
plt.show()
```

```python
sns.violinplot(data=df, x="chd", y="tobacco")

plt.xlabel("CHD")
plt.ylabel("Tobacco Consumption")
plt.title("Distribution of Tobacco Consumption by CHD")
plt.show()
```

```python
corr = df[["sbp", "obesity", "age", "ldl"]].corr()

print(corr)
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation between SBP, Obesity, Age and LDL")
plt.show()
```

```python
bins = [0, 15, 35, 55, float("inf")]
labels = ["young", "adults", "mid", "old"]

df["agegroup"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels,
    right=False
)

print(df[["age", "agegroup"]].head())
```

```python
print(df["agegroup"].value_counts())
```

```python
chd_age = df.groupby("agegroup", observed=False)["chd"].sum()

print(chd_age)
```

```python
sns.boxplot(data=df, x="agegroup", y="ldl")

plt.xlabel("Age Group")
plt.ylabel("LDL")
plt.title("Distribution of LDL across Age Groups")
plt.show()
```
