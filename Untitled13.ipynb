```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
image_path = "straight_lines.jpeg"
img = cv2.imread(image_path)
if img is None:
    print("Image not found!")
    print("Current working directory:",os.getcwd())
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray,50,150)
    lines = cv2.HoughLinesP(
        edges,
        rho = 1,
        theta = np.pi / 180,
        threshold = 50,
        minLineLength = 50,
        maxLineGap = 10
    )
    result = img_rgb.copy()
    if lines is not None:
        for line in lines:
            x1, y1, x2, y2 = line[0]
            cv2.line(result, (x1, y1), (x2, y2), (255, 0, 0), 2)

        print("Number of detected lines:", len(lines))
    else:
        print("No straight lines detected.")


    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(edges, cmap="gray")
    plt.title("Canny Edge Detection")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(result)
    plt.title("Hough Line Transform")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# 1. Load image
path = "computervision/straightline.jpeg"

if not os.path.exists(path):
    path = "straightline.jpeg"

image = cv2.imread(path)

if image is None:
    print("Image not loaded!")
    print("Current folder:", os.getcwd())
    print("Check the image path and filename.")

else:
    print("Image loaded:", image.shape)

    # 2. Convert to grayscale using a fresh variable
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray_image = np.ascontiguousarray(gray_image, dtype=np.uint8)

    print("Grayscale type:", type(gray_image))
    print("Grayscale shape:", gray_image.shape)

    # 3. Harris Corner Detection
    harris_response = cv2.cornerHarris(
        np.float32(gray_image),
        2,  # blockSize
        3,  # ksize
        0.04
    )

    harris_response = cv2.dilate(harris_response, None)

    harris_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB).copy()
    harris_image[harris_response > 0.01 * harris_response.max()] = [255, 0, 0]

    # 4. Shi-Tomasi Corner Detection
    shi_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB).copy()

    corners = cv2.goodFeaturesToTrack(
        gray_image,
        maxCorners=100,
        qualityLevel=0.01,
        minDistance=10
    )

    if corners is not None:
        for point in corners:
            x, y = point.ravel().astype(int)
            cv2.circle(shi_image, (x, y), 4, (0, 255, 0), -1)
        print("Shi-Tomasi corners:", len(corners))

    # 5. Display results
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(harris_image)
    plt.title("Harris Corners")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(shi_image)
    plt.title("Shi-Tomasi Corners")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
