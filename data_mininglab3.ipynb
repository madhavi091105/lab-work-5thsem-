```python
import numpy as np
print("=" * 50)
print("   Numpy complete demo")
print("="*50)
```

```python
print("\n1.Array Creation")
arr1 = np.array([10,20,30,40,50])
print("1D Array:",arr1)
arr2 = np.array([[1,2,3],[4,5,6]])
print("2D Array:\n",arr2)
print("Zeros:\n",np.zeros((2,3)))
print("Ones:\n",np.ones((2,3)))
print("Range:",np.arange(1,10,2))
print("Linspace:",np.linspace(0,10,5))
```

```python
print("\n2. ARRAY PROPERTIES")

print("Shape:", arr2.shape)
print("Number of dimensions:", arr2.ndim)
print("Size:", arr2.size)
print("Data type:", arr2.dtype)
```

```python
print("\n3. INDEXING")

print("First element:", arr1[0])
print("Last element:", arr1[-1])
print("Element [1][2]:", arr2[1][2])
```

```python
print("\n4. SLICING")

print("First 3 elements:", arr1[:3])
print("Last 2 elements:", arr1[-2:])
print("2D first row:", arr2[0, :])
print("2D second column:", arr2[:, 1])
```

```python
print("\n5. RESHAPING")

a = np.arange(1, 13)

print("Original:", a)

matrix = a.reshape(3, 4)

print("3 x 4 Matrix:\n", matrix)

print("Flatten:", matrix.flatten())
```

```python
print("\n6. ARRAY OPERATIONS")

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
```

```python
print("\n7. BROADCASTING")

a = np.array([10, 20, 30])

print("Original:", a)
print("Add 5:", a + 5)
print("Multiply by 2:", a * 2)
```

```python
print("\n8. MATHEMATICAL FUNCTIONS")

a = np.array([1, 4, 9, 16, 25])

print("Square root:", np.sqrt(a))
print("Square:", np.square(a))
print("Exponential:", np.exp([1, 2, 3]))
print("Log:", np.log(a))


```

```python
print("\n9. AGGREGATION")

marks = np.array([78, 85, 92, 67, 88])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))
```

```python
print("\n10. AXIS OPERATIONS")

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Matrix:\n", matrix)

print("Row-wise sum:", np.sum(matrix, axis=1))
print("Column-wise sum:", np.sum(matrix, axis=0))

print("Row-wise mean:", np.mean(matrix, axis=1))
print("Column-wise mean:", np.mean(matrix, axis=0))

```

```python
print("\n11. BOOLEAN FILTERING")

marks = np.array([45, 78, 92, 56, 88, 34])

print("Marks > 60:", marks[marks > 60])
print("Marks < 50:", marks[marks < 50])
```

```python
print("\n12. WHERE")

marks = np.array([45, 78, 92, 56, 88, 34])

result = np.where(marks >= 60, "Pass", "Fail")

print(result)
```

```python
print("\n13. SORTING")

arr = np.array([50, 20, 80, 10, 40])

print("Original:", arr)
print("Sorted:", np.sort(arr))
print("Sorted indices:", np.argsort(arr))
```

```python
print("\n14. UNIQUE")

arr = np.array([10, 20, 10, 30, 20, 40, 30])

print("Unique values:", np.unique(arr))
```

```python
print("\n15. CONCATENATION")

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Concatenated:", np.concatenate((a, b)))

```

```python
print("\n16. STACKING")

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vertical stack:")
print(np.vstack((a, b)))

print("Horizontal stack:")
print(np.hstack((a, b)))
```

```python
print("\n17. RANDOM NUMBERS")

print("Random numbers:")
print(np.random.rand(5))

print("Random integers:")
print(np.random.randint(1, 100, 5))
```

```python
print("\n18. RANDOM MATRIX")

random_matrix = np.random.randint(1, 10, (3, 3))

print(random_matrix)
```

```python
print("\n19. LINEAR ALGEBRA")

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

print("Matrix multiplication:")
print(np.matmul(A, B))

print("Transpose of A:")
print(A.T)

print("Determinant of A:")
print(np.linalg.det(A))

print("Inverse of A:")
print(np.linalg.inv(A))
```

```python
print("\n20. DOT PRODUCT")

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Dot product:", np.dot(a, b))
```

```python
print("\n21. COPY")

a = np.array([1, 2, 3])

b = a.copy()

b[0] = 100

print("Original:", a)
print("Copy:", b)
```

```python
import pandas as pd
import numpy as np

print("=" * 60)
print("              PANDAS COMPLETE DEMO")
print("=" * 60)
```

```python
print("\n1. CREATING DATAFRAME")

data = {
    "Name": ["Aman", "Riya", "Rahul", "Neha", "Karan", "Priya"],
    "Age": [20, 21, 19, 22, 20, 21],
    "Department": ["CSE", "AI", "CSE", "ECE", "AI", "CSE"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Delhi"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("\n2. SERIES")

marks = pd.Series([85, 92, 76, 88, 67])

print(marks)
```

```python
print("\n3. BASIC INFORMATION")

print("Shape:", df.shape)
print("Columns:", df.columns)
print("Index:", df.index)
print("Data Types:")
print(df.dtypes)
```

```python
print("\n4. HEAD AND TAIL")

print("First 3 rows:")
print(df.head(3))

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
print("\n5. INFO")

df.info()
```

```python
print("\n6. STATISTICAL SUMMARY")

print(df.describe())
```

```python
print("\n7. SELECTING COLUMNS")

print("Names:")
print(df["Name"])

print("\nMarks:")
print(df["Marks"])

print("\nMultiple columns:")
print(df[["Name", "Marks", "Department"]])
```

```python
print("\n8. SELECTING ROWS")

print("First row:")
print(df.iloc[0])

print("\nFirst three rows:")
print(df.iloc[:3])
```

```python
print("\n9. LOC")

print(df.loc[0, "Name"])

print("\nRows where marks > 80:")
print(df.loc[df["Marks"] > 80])
```

```python
print("\n10. FILTERING")

print("Students with marks > 80:")
print(df[df["Marks"] > 80])

print("\nStudents from Delhi:")
print(df[df["City"] == "Delhi"])

print("\nAI students with marks > 80:")
print(df[(df["Department"] == "AI") & (df["Marks"] > 80)])
```

```python
print("\n11. ADDING COLUMN")

df["Passed"] = df["Marks"] >= 40

print(df)
```

```python
print("\n12. CALCULATED COLUMN")

df["Bonus"] = df["Marks"] + 5

print(df)
```

```python
print("\n13. DROP COLUMN")

df = df.drop("Bonus", axis=1)

print(df)
```

```python
print("\n14. SORTING")

print("Sorted by Marks:")
print(df.sort_values("Marks"))

print("\nSorted by Marks descending:")
print(df.sort_values("Marks", ascending=False))
```

```python
print("\n15. SORTING BY MULTIPLE COLUMNS")

print(df.sort_values(["Department", "Marks"]))
```

```python
print("\n16. UNIQUE VALUES")

print("Departments:")
print(df["Department"].unique())

print("\nCities:")
print(df["City"].unique())
```

```python
print("\n17. VALUE COUNTS")

print("Students per department:")
print(df["Department"].value_counts())

print("\nStudents per city:")
print(df["City"].value_counts())
```

```python
print("\n18. STATISTICS")

print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
print("Total Marks:", df["Marks"].sum())
print("Median Marks:", df["Marks"].median())
print("Standard Deviation:", df["Marks"].std())
```

```python
print("\n19. GROUPBY")

print("Average marks by department:")

print(
    df.groupby("Department")["Marks"].mean()
)
```

```python
print("\n20. GROUPBY WITH MULTIPLE FUNCTIONS")

print(
    df.groupby("Department")["Marks"].agg(
        ["mean", "max", "min", "count"]
    )
)
```

```python
print("\n21. MISSING VALUES")

data_missing = {
    "Name": ["Aman", "Riya", "Rahul", "Neha"],
    "Marks": [85, np.nan, 76, 90],
    "Age": [20, 21, np.nan, 22]
}

df_missing = pd.DataFrame(data_missing)

print(df_missing)

print("\nMissing values:")
print(df_missing.isnull())

print("\nNumber of missing values:")
print(df_missing.isnull().sum())
```

```python
print("\n22. FILL MISSING VALUES")

df_missing["Marks"] = df_missing["Marks"].fillna(
    df_missing["Marks"].mean()
)

df_missing["Age"] = df_missing["Age"].fillna(
    df_missing["Age"].mean()
)

print(df_missing)
```

```python
print("\n23. DROP MISSING VALUES")

temp = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Marks": [80, np.nan, 90]
})

print("Before:")
print(temp)

print("\nAfter:")
print(temp.dropna())
```

```python
print("\n24. DUPLICATES")

duplicate_df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Aman", "Rahul"],
    "Marks": [80, 90, 80, 75]
})

print("Duplicate rows:")
print(duplicate_df.duplicated())

print("\nAfter removing duplicates:")
print(duplicate_df.drop_duplicates())
```

```python
print("\n25. RENAME COLUMNS")

df2 = df.rename(columns={
    "Marks": "Score",
    "Name": "Student_Name"
})

print(df2)
```

```python
print("\n26. APPLY FUNCTION")

df["Marks_Double"] = df["Marks"].apply(lambda x: x * 2)

print(df[["Name", "Marks", "Marks_Double"]])
```

```python
print("\n27. REPLACE")

df3 = df.copy()

df3["Department"] = df3["Department"].replace(
    {"AI": "Artificial Intelligence"}
)

print(df3[["Name", "Department"]])
```

```python
print("\n28. RESET INDEX")

sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df.reset_index(drop=True))
```

```python
print("\n29. CONCAT")

df_a = pd.DataFrame({
    "Name": ["Aman", "Riya"],
    "Marks": [80, 90]
})

df_b = pd.DataFrame({
    "Name": ["Rahul", "Neha"],
    "Marks": [75, 85]
})

combined = pd.concat([df_a, df_b], ignore_index=True)

print(combined)
```

```python
print("\n30. MERGE")

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Aman", "Riya", "Rahul"]
})

marks_data = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 92, 78]
})

merged = pd.merge(students, marks_data, on="ID")

print(merged)
```

```python
print("\n31. PIVOT TABLE")

print(
    df.pivot_table(
        values="Marks",
        index="Department",
        aggfunc="mean"
    )
)

```
