```python
import cv2
import numpy as np
```

```python
X = float(input("Enter X: "))
Y = float(input("Enter Y: "))
Z = float(input("Enter Z: "))
```

```python
fx = float(input("Enter fx: "))
fy = float(input("Enter fy: "))
cx = float(input("Enter cx: "))
cy = float(input("Enter cy: "))

           
```

```python
if Z == 0:
    print("Error: Z cannot be zero.")
```

```python
u = fx * (X / Z) + cx
v = fy * (Y /Z) + cy
```

```python
K = np.array([[fx, 0, cx],[0,fy,cy],[0,0,1]])
```

```python
print("RESULTS")

print("3D Point:")
print(f"X = {X}")
print(f"Y = {Y}")
print(f"Z = {Z}")

print("\n2D Image Coordinates:")
print(f"u = {u:.2f}")
print(f"v = {v:.2f}")

print("\nCamera Intrinsic Matrix:")
print(K)
```

```python
import cv2
import numpy as np
```

```python
img = np.zeros((500,700,3),dtype = np.uint8)
cv2.putText(img,f"3D Point: ({X},{Y},{Z})",
            (30,50),
             cv2.FONT_HERSHEY_SIMPLEX,
             0.8,
             (255,255,255),
              2
)
cv2.putText( 
    img,
    f"2D Point: (u = {u:2f},v = {v:.2f})",
    (30,100),
     cv2.FONT_HERSHEY_SIMPLEX,
     0.8,
     (255,255,255),
     2
)
cv2.putText(
    img,
    "Intrinsic Matrix K:",
    (30, 160),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (255, 255, 255),
    2
)

# Matrix values
for i in range(3):
    text = "   ".join(f"{K[i, j]:.2f}" for j in range(3))

    cv2.putText(
        img,
        text,
        (50, 210 + i * 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 255, 255),
        2
    )

cv2.imshow("Camera Projection", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
