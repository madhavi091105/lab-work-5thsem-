```python

import cv2

for i in range(5):
    for backend in [cv2.CAP_DSHOW, cv2.CAP_MSMF]:
        cap = cv2.VideoCapture(i, backend)

        if cap.isOpened():
            ret, frame = cap.read()

            if ret:
                print(f"Working camera: index={i}, backend={backend}")
                cv2.imshow(f"Camera {i}", frame)
                cv2.waitKey(0)
                cv2.destroyAllWindows()

        cap.release()

cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(1, cv2.CAP_DSHOW)

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        print("Frame nahi aa raha")
        break

    cv2.imshow("DroidCam Live", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python

import cv2

cap = cv2.VideoCapture(1, cv2.CAP_DSHOW)

print("Camera opened:", cap.isOpened())

if cap.isOpened():
    ret, frame = cap.read()
    print("Frame received:", ret)

    if ret:
        print("Frame size:", frame.shape)
        cv2.imshow("DroidCam Test", frame)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

cap.release()
```
