```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
df = pd.read_csv("SAheart.csv")
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed", case=False)]
print("Dataset Shape:", df.shape)
print("Columns:", df.columns.tolist())
features = [
    col for col in [
        "sbp", "tobacco", "ldl", "adiposity",
        "typea", "obesity", "alcohol", "age"
    ]
    if col in df.columns
]
if len(features) < 2:
    raise ValueError("Dataset does not contain enough expected numerical features.")
X = df[features].apply(pd.to_numeric, errors="coerce")
X = X.fillna(X.median())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("\nFeatures used:", features)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, method in zip(axes, ["single", "complete"]):
    Z = linkage(X_scaled, method=method)
    dendrogram(
        Z,
        ax=ax,
        truncate_mode="lastp",
        p=25,
        show_leaf_counts=True
    )
    ax.set_title(f"{method.capitalize()} Linkage Dendrogram")
    ax.set_xlabel("Clusters / Samples")
    ax.set_ylabel("Distance")
plt.tight_layout()
plt.show()
results = []
for method in ["single", "complete"]:
    for k in [2, 3, 4, 5]:
        model = AgglomerativeClustering(
            n_clusters=k,
            linkage=method
        )
        labels = model.fit_predict(X_scaled)
        score = silhouette_score(X_scaled, labels)
        results.append({
            "Linkage Method": method.capitalize(),
            "Number of Clusters": k,
            "Silhouette Score": round(score, 4)
        })
results_df = pd.DataFrame(results)
print("\nClustering Evaluation Results:")
print(results_df.to_string(index=False))
best_result = results_df.loc[
    results_df["Silhouette Score"].idxmax()
]
print("\nBest Result Among Tested Combinations:")
print(best_result.to_string())
best_method = best_result["Linkage Method"].lower()
best_k = int(best_result["Number of Clusters"])
best_model = AgglomerativeClustering(
    n_clusters=best_k,
    linkage=best_method
)
best_labels = best_model.fit_predict(X_scaled)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=best_labels,
    cmap="viridis",
    s=35
)
plt.title(f"Best Tested Clustering: {best_method.capitalize()} Linkage, k={best_k}")
plt.xlabel(features[0] + " (scaled)")
plt.ylabel(features[1] + " (scaled)")
plt.colorbar(scatter, label="Cluster")
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
df = pd.read_csv("SAheart.csv")
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed", case=False)]
print("Dataset Shape:", df.shape)
features = [
    col for col in [
        "sbp", "tobacco", "ldl", "adiposity",
        "typea", "obesity", "alcohol", "age"
    ]
    if col in df.columns
]
X = df[features].apply(pd.to_numeric, errors="coerce")
X = X.fillna(X.median())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Features used:", features)
n_clusters = 3
m_value = 2.0
centers, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    data=X_scaled.T,
    c=n_clusters,
    m=m_value,
    error=0.005,
    maxiter=1000,
    seed=42
)
membership = u.T
print("\nMembership Degrees (first 10 records):")
membership_df = pd.DataFrame(
    membership,
    columns=[f"Cluster_{i}" for i in range(n_clusters)]
)
print(membership_df.head(10).round(3))
fuzzy_labels = np.argmax(membership, axis=1)
print("\nFuzzy C-Means Cluster Counts:")
print(pd.Series(fuzzy_labels).value_counts().sort_index())
print("\nFuzzy Partition Coefficient (FPC):", round(fpc, 4))
kmeans = KMeans(
    n_clusters=n_clusters,
    random_state=42,
    n_init=10
)
kmeans_labels = kmeans.fit_predict(X_scaled)
print("\nK-Means Cluster Counts:")
print(pd.Series(kmeans_labels).value_counts().sort_index())
ari = adjusted_rand_score(fuzzy_labels, kmeans_labels)
print("\nAdjusted Rand Index (Fuzzy vs K-Means):", round(ari, 4))
plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=membership.max(axis=1),
    cmap="viridis",
    s=35
)
plt.colorbar(scatter, label="Maximum Membership Degree")
plt.title("Fuzzy C-Means Clustering on SAheart")
plt.xlabel(features[0] + " (scaled)")
plt.ylabel(features[1] + " (scaled)")
plt.tight_layout()
plt.show()
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=fuzzy_labels,
    cmap="viridis",
    s=30
)
axes[0].set_title("Fuzzy C-Means: Hard Assignments")
axes[0].set_xlabel(features[0] + " (scaled)")
axes[0].set_ylabel(features[1] + " (scaled)")

axes[1].scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=kmeans_labels,
    cmap="viridis",
    s=30
)
axes[1].set_title("K-Means Assignments")
axes[1].set_xlabel(features[0] + " (scaled)")
axes[1].set_ylabel(features[1] + " (scaled)")
plt.tight_layout()
plt.show()
print("\nEffect of Fuzziness Parameter:")
for m in [1.5, 2.0, 3.0]:
    _, u_m, _, _, _, _, fpc_m = fuzz.cluster.cmeans(
        data=X_scaled.T,
        c=n_clusters,
        m=m,
        error=0.005,
        maxiter=1000,
        seed=42
    )
    max_membership = u_m.max(axis=0)
    print(
        f"m = {m}: "
        f"FPC = {fpc_m:.4f}, "
        f"Mean maximum membership = {max_membership.mean():.4f}"
    )
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("SAheart.csv")
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
numeric_cols = ["sbp", "tobacco", "ldl", "adiposity", "age"]
categorical_cols = ["famhist"]
numeric_cols = [c for c in numeric_cols if c in df.columns]
categorical_cols = [c for c in categorical_cols if c in df.columns]
data = df[numeric_cols + categorical_cols].copy()
for col in numeric_cols:
    data[col] = pd.to_numeric(data[col], errors="coerce")
    data[col] = data[col].fillna(data[col].median())
for col in categorical_cols:
    data[col] = data[col].astype("string").fillna("Unknown")
sample_size = min(100, len(data))
sample = data.sample(n=sample_size, random_state=42).reset_index(drop=True)
n = len(sample)
D = np.zeros((n, n), dtype=float)
for col in numeric_cols:
    values = sample[col].astype(float).to_numpy()
    feature_range = values.max() - values.min()
    if feature_range > 0:
        diff = np.abs(values[:, None] - values[None, :]) / feature_range
        D += diff
for col in categorical_cols:
    values = sample[col].astype(str).to_numpy()
    diff = (values[:, None] != values[None, :]).astype(float)
    D += diff
total_features = len(numeric_cols) + len(categorical_cols)
if total_features == 0:
    raise ValueError("No suitable features found in the dataset.")
D = D / total_features
S = 1 - D
print("Numerical features:", numeric_cols)
print("Categorical features:", categorical_cols)
print("\nDissimilarity matrix shape:", D.shape)
print("Similarity matrix shape:", S.shape)
print("\nDissimilarity matrix (first 5 rows/columns):")
print(pd.DataFrame(D).iloc[:5, :5].round(3))
print("\nSimilarity matrix (first 5 rows/columns):")
print(pd.DataFrame(S).iloc[:5, :5].round(3))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.heatmap(D, cmap="Reds", ax=axes[0], vmin=0, vmax=1)
axes[0].set_title("Dissimilarity Matrix")
axes[0].set_xlabel("Sample index")
axes[0].set_ylabel("Sample index")
sns.heatmap(S, cmap="Blues", ax=axes[1], vmin=0, vmax=1)
axes[1].set_title("Similarity Matrix")
axes[1].set_xlabel("Sample index")
axes[1].set_ylabel("Sample index")
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
file_path = Path("SAheart.csv")
df = pd.read_csv(file_path)
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
df.columns = df.columns.str.strip().str.lower()
print("Original dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nDataset information:")
df.info()
print("\nMissing values before cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows before cleaning:", df.duplicated().sum())
print("\nStatistical summary:")
print(df.describe(include="all"))
df = df.drop_duplicates().copy()
numeric_cols = [
    "sbp", "tobacco", "ldl", "adiposity",
    "typea", "obesity", "alcohol", "age", "chd"
]
numeric_cols = [c for c in numeric_cols if c in df.columns]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
if "famhist" in df.columns:
    df["famhist"] = (
        df["famhist"]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace({"": pd.NA, "nan": pd.NA})
    )
for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())
if "famhist" in df.columns and df["famhist"].isnull().any():
    mode_values = df["famhist"].mode()
    fill_value = mode_values.iloc[0] if not mode_values.empty else "unknown"
    df["famhist"] = df["famhist"].fillna(fill_value)
outlier_summary = {}
for col in numeric_cols:
    values = df[col].dropna()
    if len(values) == 0:
        continue
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = ((values < lower) | (values > upper)).sum()
    outlier_summary[col] = int(outlier_count)
print("\nPotential outliers per numerical column:")
print(pd.Series(outlier_summary, name="outlier_count"))
print("\nMissing values after cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows after cleaning:", df.duplicated().sum())
print("\nCleaned dataset shape:", df.shape)
print("\nCleaned statistical summary:")
print(df.describe(include="all"))
plot_cols = [
    c for c in numeric_cols
    if c != "chd" and c in df.columns
]
if plot_cols:
    df[plot_cols].hist(figsize=(14, 9), bins=20)
    plt.suptitle("Distributions After Data Cleaning")
    plt.tight_layout()
    plt.show()
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df[plot_cols])
    plt.title("Boxplots for Outlier Inspection")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
output_path = Path("computervision/SAheart_cleaned.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)
print("\nCleaned dataset saved to:", output_path)
```
