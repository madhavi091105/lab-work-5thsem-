```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Exact path
file_path = r"C:\Users\madha\computervision\bollywood.csv"

# Load dataset
df = pd.read_csv(file_path)

print(df.head())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
# Select numerical columns
X = df.select_dtypes(include=np.number).copy()

# Handle missing values
X = X.fillna(X.median())

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Features used:")
print(X.columns.tolist())

print("\nX_scaled shape:")
print(X_scaled.shape)
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_scaled)

print("Cluster Labels:")
print(labels)

print("\nCluster Centroids:")
print(kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=labels,
    cmap="viridis"
)
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200,
    color="red",
    label="Centroids"
)
plt.xlabel(X.columns[0])
plt.ylabel(X.columns[1])
plt.title("K-Means Clustering (K=2)")
plt.legend()
plt.show()
```

```python

sse = []
k_values = [2, 3, 4]
for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    sse_value = kmeans.inertia_
    sse.append(sse_value)
    print("K =", k)
    print("SSE =", sse_value)
    print("-" * 30)
sse_table = pd.DataFrame({
    "K": k_values,
    "SSE": sse
})
print("\nSSE Comparison:")
print(sse_table)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(
    k_values,
    sse,
    marker='o'
)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("SSE")
plt.title("SSE Comparison for K = 2, 3, 4")
plt.xticks([2, 3, 4])
plt.grid(True)
plt.show()
```

```python

from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
Z_single = linkage(
    X_scaled,
    method='single'
)
plt.figure(figsize=(12, 6))
dendrogram(
    Z_single,
    truncate_mode='lastp',
    p=30
)
plt.title("Hierarchical Clustering - Single Link")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
Z_complete = linkage(
    X_scaled,
    method='complete'
)
plt.figure(figsize=(12, 6))
dendrogram(
    Z_complete,
    truncate_mode='lastp',
    p=30
)
plt.title("Hierarchical Clustering - Complete Link")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
Z_average = linkage(
    X_scaled,
    method='average'
)
plt.figure(figsize=(12, 6))
dendrogram(
    Z_average,
    truncate_mode='lastp',
    p=30
)
plt.title("Hierarchical Clustering - Average Link")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

```python

from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
eps = 1.5
MinPts = 5
dbscan = DBSCAN(
    eps=eps,
    min_samples=MinPts
)
labels = dbscan.fit_predict(X_scaled)
print("DBSCAN Cluster Labels:")
print(labels)
core_indices = dbscan.core_sample_indices_
core_mask = np.zeros(
    len(X_scaled),
    dtype=bool
)
core_mask[core_indices] = True
noise_mask = labels == -1
border_mask = (
    (labels != -1)
    & (~core_mask)
)
print("\nNumber of Core Points:")
print(np.sum(core_mask))
print("\nNumber of Border Points:")
print(np.sum(border_mask))
print("\nNumber of Noise Points:")
print(np.sum(noise_mask))
print("\nCore Point Indexes:")
print(np.where(core_mask)[0])
print("\nBorder Point Indexes:")
print(np.where(border_mask)[0])
print("\nNoise Point Indexes:")
print(np.where(noise_mask)[0])
```

```python

plt.figure(figsize=(10, 7))
plt.scatter(
    X_scaled[core_mask, 0],
    X_scaled[core_mask, 1],
    s=60,
    label="Core Points"
)
plt.scatter(
    X_scaled[border_mask, 0],
    X_scaled[border_mask, 1],
    s=40,
    label="Border Points"
)
plt.scatter(
    X_scaled[noise_mask, 0],
    X_scaled[noise_mask, 1],
    s=80,
    marker="x",
    label="Noise Points"
)
plt.xlabel(X.columns[0])
plt.ylabel(X.columns[1])
plt.title("DBSCAN Clustering")
plt.legend()
plt.show()
```

```python

import skfuzzy as fuzz
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
n_clusters = 2
data = X_scaled.T
cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    data,
    c=n_clusters,
    m=2,
    error=0.005,
    maxiter=1000,
    init=None,
    seed=42
)
print("Cluster Centers (Scaled):")
print(cntr)
cntr_original = scaler.inverse_transform(cntr)
center_df = pd.DataFrame(
    cntr_original,
    columns=X.columns
)
print("\nCluster Centers (Original Scale):")
print(center_df)
membership_df = pd.DataFrame(
    u.T,
    columns=[
        "Cluster_1_Membership",
        "Cluster_2_Membership"
    ]
)
print("\nMembership Values:")
print(membership_df)
assigned_cluster = np.argmax(u, axis=0) + 1
membership_df["Assigned_Cluster"] = assigned_cluster
print("\nMembership Values + Assigned Cluster:")
print(membership_df)
print("\nFuzzy Partition Coefficient:")
print(fpc)
```

```python

plt.figure(figsize=(8, 6))
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=assigned_cluster,
    cmap="viridis",
    s=50
)
plt.scatter(
    cntr[:, 0],
    cntr[:, 1],
    marker="X",
    s=250,
    color="red",
    label="Cluster Centers"
)
plt.xlabel(X.columns[0])
plt.ylabel(X.columns[1])
plt.title("Fuzzy C-Means Clustering")
plt.legend()
plt.show()
```
