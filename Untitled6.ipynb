```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [2, 3],
    [3, 2],

    [8, 8],
    [8, 9],
    [9, 8],
    [9, 9],
    [10, 8],

    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
])
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(X)
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
print("Cluster Labels:")
print(labels)
print("\nCluster Centroids:")
print(centroids)
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    s=200
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("K-Means Clustering (K=2)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [2, 3],
    [3, 2],

    [8, 8],
    [8, 9],
    [9, 8],
    [9, 9],
    [10, 8],

    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
])
for k in [2, 3, 4]:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)

    print("\nK =", k)
    print("SSE =", kmeans.inertia_)
    print("Centroids:")
    print(kmeans.cluster_centers_)

    plt.figure()

    plt.scatter(
        X[:, 0],
        X[:, 1],
        c=kmeans.labels_
    )

    plt.scatter(
        kmeans.cluster_centers_[:, 0],
        kmeans.cluster_centers_[:, 1],
        marker='X',
        s=200
    )

    plt.title("K-Means with K = " + str(k))
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

sse = []
K_values = [2, 3, 4]

for k in K_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)

    sse.append(kmeans.inertia_)

print("SSE values:", sse)

plt.plot(K_values, sse, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("SSE")
plt.title("SSE vs K")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.cluster.hierarchy import (
    linkage,
    dendrogram
)

X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [2, 3],
    [3, 2],

    [8, 8],
    [8, 9],
    [9, 8],
    [9, 9],
    [10, 8],

    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
])

methods = [
    "single",
    "complete",
    "average"
]

for method in methods:

    Z = linkage(X, method=method)

    plt.figure(figsize=(8, 5))

    dendrogram(Z)

    plt.title(
        "Agglomerative Hierarchical Clustering - "
        + method
    )

    plt.xlabel("Data Points")
    plt.ylabel("Distance")

    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [2, 3],
    [3, 2],

    [8, 8],
    [8, 9],
    [9, 8],
    [9, 9],
    [10, 8],

    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6],

    [15, 15]      
])

eps = 1.5
minpts = 3

dbscan = DBSCAN(
    eps=eps,
    min_samples=minpts
)

labels = dbscan.fit_predict(X)

print("Cluster Labels:")
print(labels)
core_indices = dbscan.core_sample_indices_
print("\nCore Point Indices:")
print(core_indices)
print("\nCore Points:")
print(X[core_indices])
noise_indices = np.where(labels == -1)[0]
print("\nNoise Point Indices:")
print(noise_indices)
print("\nNoise Points:")
print(X[noise_indices])
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("DBSCAN Clustering")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [2, 3],
    [3, 2],

    [8, 8],
    [8, 9],
    [9, 8],
    [9, 9],
    [10, 8],

    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
])

data = X.T

c = 3


centers, membership, _, _, _, _, _ = fuzz.cluster.cmeans(
    data,
    c=c,
    m=2,
    error=0.005,
    maxiter=1000,
    init=None
)

print("Cluster Centers:")
print(centers)

print("\nMembership Values:")
print(membership)

for i in range(len(X)):

    print(
        "Point:",
        X[i],
        "Membership:",
        membership[:, i]
    )

labels = np.argmax(membership, axis=0)

print("\nCluster Labels:")
print(labels)
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker='X',
    s=200
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Fuzzy C-Means Clustering")

plt.show()
```
