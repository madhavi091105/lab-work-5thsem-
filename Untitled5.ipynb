```python
from scipy.stats import norm

mean = 38
std = 5

prob = norm.sf(50, loc=mean, scale=std)

print("Proportion spraying more than 50 liters =", prob)
```

```python
prob = norm.cdf(10, loc=mean, scale=std)

print("Proportion spraying less than 10 liters =", prob)
```

```python
prob = norm.cdf(60, loc=mean, scale=std) - norm.cdf(30, loc=mean, scale=std)

print("Proportion spraying between 30 and 60 liters =", prob)
```

```python
import numpy as np
from scipy.stats import norm
mu = 5000
x_bar = 4998.1
sigma = 1.30
n = 60
alpha = 0.05
z = (x_bar - mu) / (sigma / np.sqrt(n))
p_value = norm.cdf(z)
critical_value = norm.ppf(alpha)
print("Z statistic =", z)
print("P-value =", p_value)
print("Critical value =", critical_value)
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
import numpy as np
from scipy.stats import norm
n = 100
x = 22
p0 = 0.18
alpha = 0.05
p_hat = x / n
se = np.sqrt(p0 * (1 - p0) / n)
z = (p_hat - p0) / se
p_value = norm.cdf(z)
print("Sample proportion =", p_hat)
print("Z statistic =", z)
print("P-value =", p_value)
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
import pandas as pd
from scipy.stats import ttest_rel
data = pd.read_csv("bollywood.csv")
data = data.dropna(subset=["YoutubeLikes", "YoutubeDislikes"])
before = data["YoutubeLikes"]
after = data["YoutubeDislikes"]
t_stat, p_two_tailed = ttest_rel(before, after)
if t_stat > 0:
    p_value = p_two_tailed / 2
else:
    p_value = 1 - (p_two_tailed / 2)
alpha = 0.05
print("T-statistic =", t_stat)
print("One-tailed P-value =", p_value)
if p_value < alpha:
    print("Reject H0")
    print("There is significant evidence that YouTube likes are greater than dislikes.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence that YouTube likes are greater than dislikes.")
```
